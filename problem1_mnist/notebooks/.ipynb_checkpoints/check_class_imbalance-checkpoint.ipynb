```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0, str(PROJECT_ROOT))
```

```python
from collections import Counter

import matplotlib.pyplot as plt
import pandas as pd

from src.data import get_dataloaders
```

```python
import os
from src.utils import load_config
PROJECT_ROOT = Path.cwd().parent
config = load_config(config_path=os.path.join(PROJECT_ROOT, "configs/data.yaml"))
data_dir = config["root"]
batch_size = config["batch_size"]
num_workers = config["num_workers"]


train_loader, val_loader, test_loader = get_dataloaders(data_dir=data_dir, batch_size=1)
print(len(train_loader))
print(len(val_loader))
print(len(test_loader))
```

```python
def get_label_counts(loader):

    counter = Counter()

    for _, labels in loader:
        counter.update(labels.tolist())

    return counter


train_counts = get_label_counts(train_loader)
val_counts = get_label_counts(val_loader)
test_counts = get_label_counts(test_loader)

print("train_counts", train_counts)
print("val_counts", val_counts)
print("test_counts", test_counts)
```

```python
df = pd.DataFrame({
    "Train": pd.Series(train_counts),
    "Validation": pd.Series(val_counts),
    "Test": pd.Series(test_counts),
}).sort_index()

df.index.name = "Digit"

df
```

```python
ax = df.plot(
    kind="bar",
    figsize=(10,5)
)

ax.set_xlabel("Digit")
ax.set_ylabel("Number of Images")
ax.set_title("Class Distribution")

plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
percentage_df = df.div(df.sum(axis=0), axis=1) * 100

percentage_df.round(2)
```

```python
for split, counts in [
    ("Train", train_counts),
    ("Validation", val_counts),
    ("Test", test_counts),
]:

    values = list(counts.values())

    print(f"{split}")
    print(f"Max samples : {max(values)}")
    print(f"Min samples : {min(values)}")
    print(f"Imbalance Ratio : {max(values)/min(values):.2f}")
    print()
```

```python
from pprint import pprint 
pprint("The class distribution was analyzed for the training, validation, and test sets. The maximum-to-minimum class imbalance ratios were 1.24, 1.24, and 1.27, respectively. These values are close to 1, indicating that no class is significantly underrepresented. Therefore, the dataset can be considered approximately balanced, and techniques such as class weighting, oversampling, or undersampling were not required. Since the class frequencies are similar, standard cross-entropy loss is expected to learn all classes without introducing significant bias toward any particular digit.")
```
