# Verification of L-smoothness

### Objective

The objective is to determine whether the optimization objective satisfies the **L-smoothness** assumption required in the convergence analysis of Stochastic Gradient Descent (SGD).

### Why it cannot be formally verified

The L-smoothness assumption states that the gradient of the objective function is **Lipschitz continuous**, i.e.,

$$
\|\nabla f(x)-\nabla f(y)\| \le L\|x-y\|,\quad \forall x,y,
$$

where \(L\) is the Lipschitz constant.

For twice-differentiable functions, this is equivalent to requiring that the spectral norm of the Hessian is bounded,

$$
\|\nabla^2 f(x)\|_2 \le L.
$$

Verifying this condition requires computing or estimating the Hessian matrix (or its largest eigenvalue) over the optimization domain. For a convolutional neural network with thousands of trainable parameters, the Hessian is extremely large and computationally expensive to construct or analyze.

### Conclusion

Therefore, **L-smoothness cannot be formally verified in this work**. The experiments performed in this project (ergodicity of the input data, step-size analysis, unbiased gradient estimation, and bounded variance of stochastic gradients) provide empirical evidence supporting stable SGD optimization. However, these observations do **not** constitute a proof of L-smoothness, which requires Hessian-based analysis beyond the scope of this implementation.
